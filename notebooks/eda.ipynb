```python
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("../data/numeric_pattern.csv")
```

```python
numeric_columns = df.columns

df[numeric_columns].hist(
    figsize=(16, 12),
    bins=30
)

plt.suptitle(
    "Histograms of Numerical Features",
    fontsize=16
)

plt.tight_layout()
plt.savefig("../outputs/histogram",bbox_inches="tight")
plt.show()
```

```python
plt.figure(figsize=(16, 8))

sns.boxplot(
    data=df
)

plt.title(
    "Boxplots of Numerical Features"
)

plt.xticks(rotation=45)

plt.tight_layout()
plt.savefig("../outputs/box",bbox_inches="tight")
plt.show()
```

```python
sns.pairplot(
    df,
    diag_kind="hist"
)
plt.savefig("../outputs/pair",bbox_inches="tight")
plt.show()
```

```python
correlation_matrix = df.corr(
    numeric_only=True
)

plt.figure(figsize=(14, 10))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title(
    "Correlation Heatmap"
)

plt.tight_layout()
plt.savefig("../outputs/heat",bbox_inches="tight")
plt.show()
```
