```python
import pandas as pd
import numpy as np
```

```python
df = pd.read_csv("../data/numeric_pattern.csv")

clean_df = df.copy()
```

```python
# Replacing null values with median

for column in clean_df.columns:
    
    if clean_df[column].isnull().sum() > 0:
        
        median_value = clean_df[column].median()
        
        clean_df[column] = clean_df[column].fillna(median_value)

print(clean_df.isnull().sum())
```

```python
# Remove duplicated values

clean_df.duplicated().sum()
```

```python
clean_df = clean_df.drop_duplicates()
clean_df.duplicated().sum()
```

```python
# Removing outliers

feature_columns = [
    column
    for column in clean_df.columns
    if column != "Target"
]
```

```python
outlier_summary = []

for column in feature_columns:

    Q1 = clean_df[column].quantile(0.25)
    Q3 = clean_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = clean_df[
        (clean_df[column] < lower_bound)
        | (clean_df[column] > upper_bound)
    ]

    outlier_summary.append({
        "Feature": column,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Outlier Count": len(outliers)
    })

outlier_summary = pd.DataFrame(outlier_summary)

display(outlier_summary)
```

```python
for column in feature_columns:

    Q1 = clean_df[column].quantile(0.25)
    Q3 = clean_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    clean_df[column] = clean_df[column].clip(
        lower=lower_bound,
        upper=upper_bound
    )
```

```python
outlier_summary_after = []

for column in feature_columns:

    Q1 = clean_df[column].quantile(0.25)
    Q3 = clean_df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = clean_df[
        (clean_df[column] < lower_bound)
        | (clean_df[column] > upper_bound)
    ]

    outlier_summary_after.append({
        "Feature": column,
        "Outlier Count": len(outliers)
    })

outlier_summary_after = pd.DataFrame(outlier_summary_after)

display(outlier_summary_after)
```

```python
clean_df.to_csv("../data/clean_numeric_pattern.csv", index = False)
```
