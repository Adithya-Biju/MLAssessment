```python
import numpy as np
import pandas as pd
from faker import Faker
```

```python
SEED = 42
N_SAMPLES = 5000

fake = Faker()
Faker.seed(SEED)

rng = np.random.default_rng(SEED)
```

```python
record_ids = [
    fake.uuid4()
    for _ in range(N_SAMPLES)
]
```

```python
df = pd.DataFrame({
    "record_id": record_ids,

    "Feature_1": rng.normal(
        loc=50,
        scale=15,
        size=N_SAMPLES
    ),

    "Feature_2": rng.uniform(
        low=0,
        high=100,
        size=N_SAMPLES
    ),

    "Feature_3": rng.normal(
        loc=100,
        scale=25,
        size=N_SAMPLES
    ),

    "Feature_4": rng.uniform(
        low=10,
        high=90,
        size=N_SAMPLES
    ),

    "Feature_5": rng.normal(
        loc=0,
        scale=1,
        size=N_SAMPLES
    ),

    "Feature_6": rng.exponential(
        scale=20,
        size=N_SAMPLES
    ),

    "Feature_7": rng.normal(
        loc=500,
        scale=100,
        size=N_SAMPLES
    ),

    "Feature_8": rng.uniform(
        low=1,
        high=10,
        size=N_SAMPLES
    ),

    "Feature_9": rng.normal(
        loc=25,
        scale=5,
        size=N_SAMPLES
    ),

    "Feature_10": rng.uniform(
        low=-5,
        high=5,
        size=N_SAMPLES
    )
})
```

```python
noise = rng.normal(
    loc=0,
    scale=12,
    size=N_SAMPLES
)


df["Target"] = (
    2.8 * df["Feature_1"]
    + 0.9 * df["Feature_2"]
    + 1.5 * df["Feature_3"]
    - 1.2 * df["Feature_4"]
    + 8.0 * df["Feature_5"]
    + 0.035 * df["Feature_7"]

    # Non-linear relationship
    + 4.0 * np.square(df["Feature_8"])

    + 2.0 * df["Feature_9"]

    # Another non-linear relationship
    + 5.0 * np.sin(df["Feature_10"])

    + noise
)
```

```python
df.head()
```

```python
# null values
for column in [
    "Feature_2",
    "Feature_5",
    "Feature_9"
]:

    missing_indices = rng.choice(
        df.index,
        size=int(N_SAMPLES * 0.01),
        replace=False
    )

    df.loc[
        missing_indices,
        column
    ] = np.nan
```

```python
# Duplicate values
duplicates = df.sample(
    n=25,
    random_state=SEED
)

df = pd.concat(
    [df, duplicates],
    ignore_index=True
)
```

```python
# Outliers
outlier_indices = rng.choice(
    df.index,
    size=10,
    replace=False
)

df.loc[
    outlier_indices,
    "Feature_1"
] *= 5

df.loc[
    outlier_indices,
    "Feature_7"
] *= 3
```

```python
df = df.drop(
    columns=["record_id"]
)
```
